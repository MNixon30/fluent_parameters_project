```python
import ansys.fluent.core as pyfluent
session = pyfluent.launch_fluent(
        ui_mode='gui', 
        mode = 'solver', 
        precision='single', 
        processor_count=2,
        dimension=2)

        
```

```python
session.settings.file.start_journal(file_name=r"C:\Users\mitch\Ravens_Racing_CFD\z_Project_Test\cas_jou2")
```

```python
session.settings.file.read_mesh(file_name=r"C:\Users\mitch\Ravens_Racing_CFD\z_Project_Test\First_Test_2d.msh.h5")
```

```python
session.settings.setup.boundary_conditions.velocity_inlet['inlet'].momentum.velocity_magnitude.value = 17
```

```python
session.settings.setup.boundary_conditions.wall['bottom_wall'].momentum.wall_motion = 'Moving Wall'
```

```python
session.settings.setup.boundary_conditions.wall['bottom_wall'].momentum.speed.value = 17
```

```python
session.settings.solution.report_definitions.lift.create('downforce')
```

```python
session.settings.solution.report_definitions.lift['downforce'].force_vector = [0,-1,0]
```

```python
session.settings.solution.report_definitions.lift['downforce'].report_output_type = 'Lift Force'
```

```python
session.settings.solution.report_definitions.lift['downforce'].output_parameter = True
```

```python
session.settings.solution.report_definitions.drag.create('drag')

```

```python
session.settings.solution.report_definitions['drag'].report_output_type ='Drag Force'
```

```python
session.settings.solution.report_definitions['drag'].output_parameter = True
```

```python
session.settings.solution.report_definitions['drag'].zones = ['lower_wing', 'upper_wing']
```

```python
session.settings.solution.report_definitions['downforce'].zones = ['lower_wing', 'upper_wing']
```

```python
session.settings.solution.initialization.initialize()
```

```python
session.settings.solution.run_calculation.iter_count = 150
session.settings.solution.run_calculation.calculate()

```

```python
session.settings.file.stop_journal()
```

```python
session.exit()
```
